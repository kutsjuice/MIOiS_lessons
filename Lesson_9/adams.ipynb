```julia
using Pkg; Pkg.activate("..")
```

```julia
using Symbolics
@variables x1, x2, x3, f1, f2, f3, h, x
```

```julia
L_1 = (x - x_2) / (x_1 - x_2) * (x - x_3) /(x_1 - x_3)
```

```julia
L_2 = (x - x_1) / (x_2 - x_1) * (x - x_3) / (x_2 - x_3)
```

```julia
L_3 = (x - x_1) / (x_3 - x_1) * (x - x_2) / (x_3 - x_2)
```

```julia
PL = f_1 * L_1 + f_2 * L_2 + f_3 * L_3
```

```julia
simplify(PL)
```

```julia
subs = Dict([
    x_1 => -2*h
    x_2 => -h
    x_3 => 0]
)
Ls = simplify(substitute(PL, subs))
```

```julia
f_1_2 = (f_2 - f_1) / (x_2 - x_1)
```

```julia
f_2_3 = (f_3 - f_2) / (x_3 - x_2)
```

```julia
f_1_2_3 = (f_2_3 - f_1_2) / (x_3 - x_1)
```

```julia
simplify(f_1_2_3)
```

```julia
PN = f_1 + f_1_2 * (x - x_1) + f_1_2_3 * (x - x_1) * (x - x_2)
```

```julia
PN = substitute(PN, subs)
simplify(PN)
```
